```python
# % stop after prescription
lab_flows=["Appointment → Prescription → Lab","Appointment → Prescription → Imaging","Appointment → Prescription → Pharmacy","Appointment → Prescription → Lab + Imaging","Appointment → Prescription → Lab + Pharmacy","Appointment → Prescription → Imaging + Pharmacy"]
condition=(df["flow_name"]=="Appointment → Prescription") & (~df["flow_name"].isin(lab_flows))
df[condition]["patient_id"].nunique()/df["patient_id"].nunique()*100
```

```python
# % continue to Lab/Imaging/Pharmacy
lab_flows=["Appointment → Prescription → Lab","Appointment → Prescription → Imaging","Appointment → Prescription → Pharmacy","Appointment → Prescription → Lab + Imaging","Appointment → Prescription → Lab + Pharmacy","Appointment → Prescription → Imaging + Pharmacy"]
df[df["flow_name"].isin(lab_flows)]["patient_id"].nunique()/df["patient_id"].nunique()*100
```

```python
# % multiple downstream services
lab_flows=["Appointment → Prescription → Lab + Imaging","Appointment → Prescription → Lab + Pharmacy","Appointment → Prescription → Imaging + Pharmacy"]
df[df["flow_name"].isin(lab_flows)]["patient_id"].nunique()/df["patient_id"].nunique()*100
```

```python
#Lab frequency
lab_flows=["Appointment → Prescription → Lab","Appointment → Prescription → Lab + Imaging","Appointment → Prescription → Lab + Pharmacy"]
df[df["flow_name"].isin(lab_flows)].groupby(df["prescription_end_datetime"].dt.month)["flow_name"].count()
```

```python
# Imaging frequency
lab_flows=["Appointment → Prescription → Imaging","Appointment → Prescription → Lab + Imaging","Appointment → Prescription → Imaging + Pharmacy"]
df[df["flow_name"].isin(lab_flows)].groupby(df["prescription_end_datetime"].dt.month)["flow_name"].count()
```

```python
#Pharmacy frequency
lab_flows=["Appointment → Prescription → Pharmacy","Appointment → Prescription → Lab + Pharmacy","Appointment → Prescription → Imaging + Pharmacy"]
df[df["flow_name"].isin(lab_flows)].groupby(df["prescription_end_datetime"].dt.month)["flow_name"].count()
```

```python
#Service combinations
pattern = "Appointment → Prescription → Lab \+ Imaging|Appointment → Prescription → Lab \+ Pharmacy|Appointment → Prescription → Imaging \+ Pharmacy"
df[df["flow_name"].str.contains(pattern,na=False)]["flow_name"].value_counts()
```

```python
# Service utilization over time
monthly_service = (
    df.assign(
        month=df["appointment_start_datetime"].dt.to_period("M"),
        has_lab=df["flow_name"].str.contains("Lab", na=False),
        has_imaging=df["flow_name"].str.contains("Imaging", na=False),
        has_pharmacy=df["flow_name"].str.contains("Pharmacy", na=False)
    )
    .groupby("month")[["has_lab", "has_imaging", "has_pharmacy"]]
    .mean()
    .mul(100)
)
monthly_service
```

```python

# Proportion requiring Lab
df["has_Lab"] = df["flow_name"].str.contains("Lab", na=False)
df.groupby("prescription_doctor_name")["has_Lab"].mean().mul(100).sort_values(ascending=False)
```

```python
# Proportion requiring Imaging
df["has_Imaging"] = df["flow_name"].str.contains("Imaging", na=False)
df.groupby("prescription_doctor_name")["has_Imaging"].mean().mul(100).sort_values(ascending=False)
```

```python
#Proportion requiring Pharmacy
df["has_Pharmacy"] = df["flow_name"].str.contains("Pharmacy", na=False)
df.groupby("prescription_doctor_name")["has_Pharmacy"].mean().mul(100).sort_values(ascending=False)
```

```python
# Proportion requiring Multiple Service
df["multi_service"] = (df[["has_Lab","has_Imaging","has_Pharmacy"]].sum(axis=1)>=2)
df.groupby("prescription_doctor_name")["multi_service"].mean().mul(100).sort_values(ascending=False)
```
